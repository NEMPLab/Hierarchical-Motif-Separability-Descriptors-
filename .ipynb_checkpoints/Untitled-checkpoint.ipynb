```python
import os
import glob
from structure_feature_extractor import (
    extract_motif_features,
    analyze_motif_connectivity,
    compute_motif_weights_standardized,
    weighted_features,
    comprehensive_weighted_spatial_analysis
)
from define_structural_primitives import (
get_crystalnn_motifs,
identify_asymmetric_unit
)
from pymatgen.core import Structure
import numpy as np
from CE_feature import (
extract_CE_features_from_motifs,
rename_CE_feature_columns,
weighted_CE_vector
)
```

```python
cif_folder = '/home/gtj/modnet-master/ext_struc/cif'
cif_files = glob.glob(os.path.join(cif_folder, '*.cif'))
print(f"共找到 {len(cif_files)} 个 CIF 文件")
```

```python
import pandas as pd
all_results = []
for cif_file in cif_files:
    try:
        struct = Structure.from_file(cif_file)
        
        # 获取不等价原子和对称化结构
        inequivalent_sites, equivalent_sites, space_group_info,symmetrized_structure = identify_asymmetric_unit(struct)
        print(f"inequivalent_sites type: {type(inequivalent_sites)}")
        print(f"inequivalent_sites content: {inequivalent_sites}")
        print(f"等价位点组数: {len(equivalent_sites)}")
        print(f"不等价原子数量: {len(inequivalent_sites)}")
        print(f"约化比例: {len(inequivalent_sites)/len(struct):.3f}")
        # 定义结构基元
        structural_motifs = get_crystalnn_motifs(
            struct,
            symmetrized_structure,
            inequivalent_sites
        )
        print(f"structural_motifs type: {type(structural_motifs)}")
        print(f"structural_motifs content: {structural_motifs}")
        representative_motifs = [
             m for m in structural_motifs
              if m['central_atom']['index'] == m['central_atom']['representative_index']
         ]
        # ---补充CE特征----
        ce_features, property_names = extract_CE_features_from_motifs(structural_motifs, "/home/gaotianjiao/modnet-master/ext_struc/element_properties/ElementsProperties.xlsx")
        final_CE_vector = weighted_CE_vector(ce_features, method='hybrid')
        final_CE_vector = rename_CE_feature_columns(pd.DataFrame([final_CE_vector]), property_names).iloc[0]
        # --- 1. 基元特征 ---
        motif_features = extract_motif_features(representative_motifs)
        weights, all_keys, X_scaled, var_w, dist_w = compute_motif_weights_standardized(motif_features, method='hybrid')
        local_features = weighted_features(motif_features, weights)
        # --- 2. 基元连接特征 ---
        connectivity = analyze_motif_connectivity(structural_motifs, struct)
      

        # --- 3. 空间对称性特征 ---
        arrangement = comprehensive_spatial_analysis(structural_motifs, struct, weight_method='hybrid')
      

        all_results.append({
            'cif_file': os.path.basename(cif_file),
            'inequivalent_sites':len(inequivalent_sites),
            'CE':final_CE_vector,
            'motif_features': local_features,
            'connectivity': connectivity,
            'arrangement': arrangement
        })

        print(f"Processed: {os.path.basename(cif_file)}")

    except Exception as e:
        print(f"Error processing {cif_file}: {e}")
```

```python
print(f"共处理 {len(all_results)} 个结构")
```

```python
import numpy as np

def safe_dict_to_vector(x):
    """
    安全地将数值特征转换为向量：
    - 如果是 dict → 提取数值型字段
    - 如果是 list → 尝试展开里面的 dict（取第一个）
    - 其他类型 → 忽略
    """
    vec = []
    
    if isinstance(x, dict):
        for k, v in x.items():
            if isinstance(v, (int, float, np.integer, np.floating)):
                vec.append(float(v))
    elif isinstance(x, list):
        # 尝试处理 list 中的 dict
        for item in x:
            if isinstance(item, dict):
                vec.extend(safe_dict_to_vector(item))
    return np.array(vec, dtype=float)

# 应用到所有 CIF
motif_vectors = [safe_dict_to_vector(r['motif_features']) for r in all_results]
connectivity_vectors = [safe_dict_to_vector(r['connectivity']) for r in all_results]
arrangement_vectors = [safe_dict_to_vector(r['arrangement']) for r in all_results]
```

```python
CE = [res['CE'] for res in all_results]
```
