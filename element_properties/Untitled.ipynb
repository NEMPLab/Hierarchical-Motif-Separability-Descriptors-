```python
import pandas as pd

df1 = pd.read_excel('/home/gtj/modnet-master/ext_struc/element_properties/ElementsProperties.xlsx', sheet_name='property')

df2 = pd.read_csv('/home/gtj/modnet-master/ext_struc/element_properties/elements2.csv')

df2['symbol'] = df2['symbol'].str.strip()

df1_t = df1.set_index('Property').T
df1_t.index.name = 'symbol'
df1_t.reset_index(inplace=True)

merged = pd.merge(df1_t, df2, on='symbol', how='left')

final = merged.set_index('symbol').T

final.to_excel('/home/gtj/modnet-master/ext_struc/element_properties/ElementsProperties_merged.xlsx', sheet_name='property')
```

```python
import pandas as pd

df1 = pd.read_excel('/home/gtj/modnet-master/ext_struc/element_properties/ElementsProperties.xlsx', sheet_name='property')

df1 = df1.loc[:, ~df1.columns.duplicated()]

df2 = pd.read_csv('/home/gtj/modnet-master/ext_struc/element_properties/elements2.csv')

df2_t = df2.set_index('symbol').T
df2_t.index.name = 'Property'
df2_t.reset_index(inplace=True)

common_elements = [col for col in df1.columns if col in df2_t.columns]
df2_t = df2_t[['Property'] + common_elements]

merged = pd.concat([df1, df2_t], ignore_index=True)

merged.to_excel('/home/gtj/modnet-master/ext_struc/element_properties/ElementsProperties_extended.xlsx', sheet_name='property', index=False)
```
