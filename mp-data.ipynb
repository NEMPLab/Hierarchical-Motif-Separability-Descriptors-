```python
import pandas as pd
from mp_api.client import MPRester
from tqdm import tqdm  # 导入进度条库

MAPI_KEY = "7PYF7c0DvFXufDMo1DL6yLZQbv7ZMpNp"

with MPRester(MAPI_KEY) as mpr:
    docs = mpr.materials.summary.search(
        fields=["material_id", "formula_pretty", "formation_energy_per_atom", "band_gap", "volume"]
    )
    print(f"共获取 {len(docs)} 条数据。")

    data_list = []
    
    # 这里加上 tqdm，会显示一个处理进度条
    for d in tqdm(docs, desc="处理数据中"):
        data_list.append({
            "mp_id": d.material_id,
            "formula": d.formula_pretty,
            "energy": d.formation_energy_per_atom,
            "band_gap": d.band_gap,
            "volume": d.volume
        })

    print("正在写入 CSV 文件...")
    df = pd.DataFrame(data_list)
    
    # 确保文件夹存在（可选，防止报错）
    import os
    output_path = "/home/gaotianjiao/modnet-master/ext_struc/mp_data.csv"
    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    
    df.to_csv(output_path, index=False)
    
    print(f"成功！数据已保存到 {output_path}")
```
